```python
import matplotlib.pyplot as plt
```

Parametrization of Klein bottle in $\mathbb{R}^4$ without self-intersection

```python
def coords(u,v):
    a, b = 1, 1
    x = (a+b*np.cos(v))*np.cos(u)
    y = (a+b*np.cos(v))*np.sin(u)
    z = b*np.sin(v)*np.cos(u/2)
    t = b*np.sin(v)*np.sin(u/2)
    return np.array([x,y,z,t])
```

```python
us = np.random.uniform(0,2*np.pi,1_000)
vs = np.random.uniform(0,2*np.pi,1_000)
```

```python
x = np.array([coords(u,v) for u,v in zip(us,vs)])
```

```python
x.shape
```

```python
diagrams2 = ripser(x, coeff=2, maxdim=2)['dgms']

plot_diagrams(diagrams2)
```

```python
plt.title('Z_2')
plot_diagrams(diagrams2)
```

```python
diagrams3 = ripser(x, coeff=3, maxdim=2)['dgms']

plot_diagrams(diagrams3)
```

```python
plt.title('Z_3')
plot_diagrams(diagrams3)
```
